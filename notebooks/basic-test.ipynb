```python
#텐서플로우 라이브러리 불러오기
import tensorflow as tf

# 바이트로 문자를 저장
hello = tf.constant('Hello, TensorFlow!')
print(hello.numpy())   # 바로 값 출력

a = tf.constant(10)
b = tf.constant(32)
c = tf.add(a, b)  # a + b 로도 쓸 수 있음
print(c.numpy())  # 결과값 출력

#numpy -> numberic python
print(a.numpy(), b.numpy(), c.numpy())
 
```
